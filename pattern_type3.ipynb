```python
n = 4
val = 1
for i in range(n):
    for j in range(n-1-i):
        print(" ", end=" ")
    for k in range(2*i+1):
        if i % 2 == 0:
            print(val, end=" ")
        else:
            print("*", end=" ")
    print() 
    if i % 2 == 0:
        val += 1
```

```python
n = 4
val = 1
p = True

for i in range(n):
    for j in range(n-1-i):
        print(" ", end=" ")
    for k in range(2*i+1):
        if p:
            print(val, end=" ")
            val += 1
            p = False
        else:
            print("*", end=" ")
            p = True
    print()
```

```python
n = 4
val = 1

for i in range(n):
    ch = ord("A")
    for j in range(n-1-i):
        print(" ", end=" ")
    for k in range(2*i+1):
        if i % 2 == 0:
            print(val, end=" ")
        else:
            print(chr(ch), end=" ")
            ch += 1
    print()
    if i % 2 == 0: val += 1
```
