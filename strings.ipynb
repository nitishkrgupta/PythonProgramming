```python
# Wap to check the given strings are anagram strings or not
def bubble_sort(s1):
    l1 = list(s1)
    for i in range(len(l1)):
        for j in range(i+1, len(l1)):
            if l1[i] > l1[j]:
                l1[i], l1[j] = l1[j], l1[i]
    return "".join(l1)

# s1 = 'silent'
# s2 = 'listen'
s1 = 'the classroom'
s2 = 'school master'

if bubble_sort(s1) == bubble_sort(s2):
    print("Anagram strings")
else:
    print("Not an Anagram strings")
```

```python
# create a class by the name vending machine and it should contains methods like
# add_items
# display_items()
# update_items()
# total_bill()

# when the object is created an empty cart should create automatically and we need to pass dictionary items as follows

class Vending_machine:
    def __init__(self, all_products):
        self.cart_items = {}
        self.all_products = all_products
    
    def add_items(self, item_name, qty):
        if item_name not in self.all_products:
            print(f"{item_name} is not available")
        else:
            self.cart_items[item_name] = qty
            print(f"{item_name} added to the cart")

    def display_items(self):
        if len(self.cart_items) <= 0:
            print("Cart is empty")
        else:
            for item_name, qty in self.cart_items.items():
                print(f"{item_name} : {qty}")

    def update_items(self, item_name, qty):
        if item_name not in self.cart_items:
            print(f"{item_name} not in the cart")
        else:
            if qty <= 0:
                del self.cart_items[item_name]
                print(f"{item_name} is removed from the cart")
            else:
                self.cart_items[item_name] = qty
                print(f"{item_name} updated in the cart")

    def total_bill(self):
        if len(self.cart_items) <= 0:
            print("Cart is empty")
        else:
            total = 0
            for item_name, qty in self.cart_items.items():
                total_i = 0
                price = self.all_products[item_name]
                total += qty * price
                total_i += qty * price

                print(f"{item_name} : {qty} * {price} = {total_i}")
            print("----------------------------")
            print(f"Total = {total}")

all_products = {'Bingo': 20,
                'cola' : 35,
                'oreo' : 30,
                'jimjam' : 40,
                'diet coke' : 60,
                'chikki' : 10,
                'red bull' : 200}
obj = Vending_machine(all_products)
obj.display_items()
obj.add_items('Bingo', 8)
obj.add_items('cola', 15)
obj.add_items('red bull', 12)
print()
obj.display_items()
print()
obj.update_items('cola', 5)
print()
obj.display_items()
print()
obj.total_bill()
```

```python
import random
class Flight_management:
    def __init__(self, flights):
        self.flights = flights
    
    def generate_tickets(self, flight_id, total_tickets):
        if flight_id not in self.flights['flight_id']:
            print("Flight not exists")
        else:
            for i in range(1, total_tickets+1):
                ticket = ""
                ticket += str(i) + self.flights['flight_name'][:3].upper() + self.flights['starting'][:3].upper() + str(random.randint(11111, 99999)) + self.flights['dest'][:3].upper()
                print(ticket)

    def get_total_cost(self, flight_id, total_tickets):
        if flight_id not in self.flights['flight_id']:
            print("Flight not exists")
        else:
            for ticket in range(1, total_tickets+1):
                total = 0
                total += self.flights['Price'] * total_tickets + ((self.flights['Price'] * total_tickets) * 0.02) + ((self.flights['Price'] * total_tickets) * 0.065)
            print("-----------------------")
            print("Total Price = ", total)
            print()

    def book_ticket(self, flight_id, total_tickets):
        if flight_id not in self.flights['flight_id']:
            print("Flight not available")
        else:
            if total_tickets > self.flights['avail_seats']:
                print(f"{total_tickets} Not available")
            else:
                print(f"{total_tickets} Ticket booked succesfully")
                self.flights['avail_seats'] = self.flights['total_seats'] - total_tickets
                print("avail_seats = ", self.flights['avail_seats'])

flights = {'flight_id' : '1ABC234', 
           'flight_name' : 'Air India', 
           'starting' : 'Banglore', 
           'dest' : 'Dubai', 
           'total_seats' : 145, 
           'avail_seats' : 145, 
           'Price' : 22000}

f = Flight_management(flights)
f.generate_tickets('1ABC234', 5)
f.get_total_cost('1ABC234', 5)
f.book_ticket('1ABC234', 5)
```

```python
# wap to convert the given integer to roman number
# the integers must be greater than 1 less than thousand
roman_pairs = [(1, 'I'), 
               (4, 'IV'),
               (5, 'V'),
               (9, 'IX'),
               (10, 'X'),
               (40, 'XL'),
               (50, 'L'),
               (90, 'XC'),
               (100, 'C'),
               (400, 'CD'),
               (500, 'D'),
               (900, 'CM'),
               (1000, 'M')]

def int_to_roman(num):
    if num >= 1 and num <= 1000:
        res = ""
        for val, rom in roman_pairs:
            while num >= val:
                res += rom
                num -= val
        return res
    else:
        return "Enter number less than 1000 and greater than 1"

num = int(input("Enter number (1, 1000) to convert : "))
print(int_to_roman(num))
```
