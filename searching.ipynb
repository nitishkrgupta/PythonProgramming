```python
# linear search algorithm

l1 = [10, 2, 5, 8, 3, 4]
key = 5
f = False
for i in range(len(l1)):
    if key == l1[i]:
        print(f"{key} is found at {i}")
        f = True
        break
if not f:
    print(f"{key} is not found")
```

```python
# finding nth occurence of key
l1 = [10, 2, 5, 8, 5, 10, 100]
key = 5
n = 3 # nth occurence
f = False
count = 0
for i in range(len(l1)):
    if key == l1[i]:
        count += 1
        if count == n:
            print(f"{key} is found at {i}")
            break
else:
    print(f"{key} is not found")
```

```python
# binary search
l1 = [10, 20, 30, 40, 50, 60, 70]
key = 11
first = 0
last = len(l1) -1
# found = False

while first <= last:
    mid = (first + last) // 2
    if key < l1[mid]:
        last = mid - 1
    elif key > l1[mid]:
        first = mid + 1
    else:
        print(f"{key} is found at index {mid}")
        # found = True
        break
else:
    print("Value not found")
```
