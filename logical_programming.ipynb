```python
def fact(n):
    prod = 1
    for i in range(1, n+1):
        prod = prod*i
    return prod

print(fact(5))
```

```python
class Factorial:
    def fact(self, n):
        prod = 1
        for i in range(1, n+1):
            prod *= i
        return prod
    
obj = Factorial()
obj.fact(5)
```

```python
# WAP to check the given number is strong number or not
n = 145
temp = n
sum = 0

while n!=0:
    r = n % 10
    fact = 1
    for i in range(1, r+1):
        fact = fact * i
    sum = sum + fact
    n = n // 10
if sum == temp:
    print(f"{temp} is Strong number")
else:
    print(f"{temp} Not a Strong number")
```

```python
# decimal to octal
n = 33
oct = 0
temp = 1

while n !=0:
    r = n % 8
    oct = oct + temp * r
    temp = temp * 10
    n = n // 8

print(oct)
```

```python
# octal to decimal
n = 123
exp = 0
sum = 0
while n != 0:
    r = n % 10
    sum = sum + r * 8**exp
    n = n // 10
    exp += 1
print(sum)
```
