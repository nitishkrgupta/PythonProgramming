```python
# wap to accept rows and columns using user input and display the elements in the form of matrix multiplication
row = int(input("Enter the number of rows : "))
col = int(input("Enter the number of columns : "))
matrix1 = [[int(input("Enter the values")) for col in range(col)] for row in range(row)]
print("Matrix1")
for i in range(row):
    for j in range(col):
        print(matrix1[i][j], end=" ")
    print()

matrix2 = [[int(input("Enter the values")) for col in range(col)] for row in range(row)]
print("Matrix2")
for i in range(row):
    for j in range(col):
        print(matrix2[i][j], end=" ")
    print()

result = [[0 for col in range(col)] for row in range(row)]

for i in range(row): # len(matrix1)
    for j in range(col): # len(matrix2[0])
        for k in range(col): # len(matrix2)
            result[i][j] += matrix1[i][k] * matrix2[k][j]

print("Result")
for i in range(row):
    for j in range(col):
        print(result[i][j], end=" ")
    print()
```

```python
# difference of sum of two diagonals
row = int(input("Enter the number of rows : "))
col = int(input("Enter the number of columns : "))
matrix1 = [[int(input("Enter the values")) for col in range(col)] for row in range(row)]
print("Matrix1")
for i in range(row):
    for j in range(col):
        print(matrix1[i][j], end=" ")
    print()

sum1 = 0
for i in range(row):
    for j in range(col):
        if i == j:
            sum1 += matrix1[i][j]
print("sum of first diagonal : ",sum1)

sum2 = 0
for i in range(row):
    for j in range(col):
        if i + j == (row - 1):
            sum2 += matrix1[i][j]

print("sum of second diagonal : ",sum2)
print(sum2 - sum1)
```

```python
# difference of sum of two diagonals
row = int(input("Enter the number of rows : "))
col = int(input("Enter the number of columns : "))
matrix1 = [[int(input("Enter the values")) for col in range(col)] for row in range(row)]
print("Matrix1")
for i in range(row):
    for j in range(col):
        print(matrix1[i][j], end=" ")
    print()
d1 = 0
for i in range(row):
    for j in range(col):
        if i == j:
            d1  += matrix1[i][j]
        if i+j == (len(matrix1) - 1):
            d1 -= matrix1[i][j]
print(abs(d1))
```

```python
# maximum element present in the list

row = int(input("Enter the number of rows : "))
col = int(input("Enter the number of columns : "))
matrix1 = [[int(input("Enter the values")) for col in range(col)] for row in range(row)]
print("Matrix1")
for i in range(row):
    for j in range(col):
        print(matrix1[i][j], end=" ")
    print()

m1 = matrix1[0][0]
for i in range(len(matrix1)):
    for j in range(len(matrix1)):
        if matrix1[i][j] > m1:
            m1 = matrix1[i][j]
print("maximum element is : ",m1)
```

```python
# sum of each row and sum of each column

row = int(input("Enter the number of rows : "))
col = int(input("Enter the number of columns : "))
matrix1 = [[int(input("Enter the values")) for col in range(col)] for row in range(row)]
print("Matrix1")
for i in range(row):
    for j in range(col):
        print(matrix1[i][j], end=" ")
    print()


for i in range(len(matrix1)):
    row_sum = 0
    for j in range(len(matrix1)):
        row_sum += matrix1[i][j]
    print(f"sum of row {i} : {row_sum}")
    col_sum = 0
    for k in range(len(matrix1)):
        col_sum += matrix1[k][i]
    print(f"sum of col {i} : {col_sum}")
```

```python
row = int(input("Enter the number of rows : "))
col = int(input("Enter the number of columns : "))
matrix1 = [[int(input("Enter the values")) for col in range(col)] for row in range(row)]
print("Matrix1")
for i in range(row):
    for j in range(col):
        print(matrix1[i][j], end=" ")
    print()

print("Result")
result = [[0 for col in range(col)] for row in range(row)]
result[0][0] = matrix1[0][0]
for i in range(len(matrix1)):
    for j in range(len(matrix1)):
        if j == 0:
            result[i][j] = matrix1[i][j]
        else:
            if matrix1[i][j-1] <= matrix1[i][j]:
                result[i][j] = matrix1[i][j] - matrix1[i][j-1]

for i in range(len(matrix1)):
    for j in range(len(matrix1)):
        print(result[i][j], end=" ")
    print()
```
