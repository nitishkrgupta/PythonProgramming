```python
# decimal to binary using recursion
def d2b(n):
    if n == 0:
        return ""
    else:
        return d2b(n//2) + str(n%2)
print(d2b(11))
```

```python
# binary to decimal
n = 1010
exp = 0
res = 0
while n != 0:
    r = n % 10
    if r == 0:
        exp += 1
    else:
        res = res + 2**exp
        exp +=1
    n = n // 10
print(res)
```

```python
# find the start and end index of a target value from a list

def first_last_idx(l1, target):
    first_idx = -1
    last_idx = -1

    for i in range(len(l1)):
        if target == l1[i]:
            if first_idx == -1:
                first_idx = i
            last_idx = i

    return first_idx, last_idx

target = int(input("Enter target : "))
l1 = [1, 2, 3, 7, 11, 3, 5, 6, 3, 12, 14]
l2 = [11, 7, 8, 6, 2]
l3 = [-1, 8, 9, 10]
print(first_last_idx(l1, target))
```

```python
# sum of prime numbers in a list
def prime(num):
    if num == 1:
        return False
    
    count = 0
    for i in range(2, num):
       if num % i == 0:
           count += 1
    if count == 0:
        return True
    else:
        return False
l1 = [2, 1, 3, 6, 5, 1, 4]
sum = 0
for i in range(len(l1)):
    if prime(l1[i]):
        sum = sum + l1[i]
sum
```

```python
# wap to check second list is a child of first list or not
l1 = [1, 2, 3, 4, 5, 6, 7, 8]
l2 = [6, 7, 8]
p = False
for i in range(len(l1)-2):
    if l1[i] in l2 and l1[i+1] in l2 and l1[i+2] in l2:
        p = True
if p:
    print(f"{l2} is child of {l1}")
else:
    print(f"{l2} is not child of {l1}")
```

```python
def func(l1, l2):
    for i in range(len(l1)):
        if l1[i:i+len(l2)] == l2:
            return True
    return False

func([1, 2, 3, 4, 5, 6, 7, 8], [6, 7, 8])
if func(l1, l2):
    print(f"{l2} is child of {l1}")
else:
    print(f"{l2} is not child of {l1}")
```

```python
# display the union of two list

l1 = [1, 2, 3, 4, 5, 6]
l2 = [5, 6, 7, 8, 1]

for i in l2:
    if i not in l1:
        l1 += [i]
print(l1)
```

```python
def union(l1, l2):
    for i in l2:
        if i not in l1:
            l1 += [i]
    return l1
print(union([1, 2, 3, 4, 5, 6], [5, 6, 7, 8, 1]))
```

```python
def intersection(l1, l2):
    new_list = []
    for i in l1:
        if i in l2:
            new_list += [i]
    return new_list
print(intersection([1, 2, 3, 4, 5, 6], [5, 6, 7, 8, 1]))
```
