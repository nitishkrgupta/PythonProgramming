```python
# program to check entered number is even or odd

n = int(input("Enter number : "))
```

```python
# 1. Using modulus (%)
if n % 2 == 0:
    print(f'{n} is even')
else:
    print(f'{n} is odd')
```

```python
# 2. using floor division (//)

if (n // 2) * 2 == n:
    print(f'{n} is even')
else:
    print(f'{n} is odd')
```

```python
# 3. Using bitwise and (&)

if (n & 1) == 0:
    print(f'{n} is even')
else:
    print(f'{n} is odd')
```

```python
# 4. Using bitwise or (|)

if (n | 1) == n:
    print(f'{n} is odd')
else:
    print(f'{n} is even')
```

```python
# 5. Using left shift and right shift

if ((n >> 1) << 1) == n:
    print(f'{n} is even')
else:
    print(f'{n} is odd')
```

```python
# 6. Without using conditional statement

l1 = ['even', 'odd']
print(l1[n % 2])
```
