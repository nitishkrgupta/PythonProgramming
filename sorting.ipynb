```python
# bubble sort for ascending order
# l1 = [5, 4, 3, 2, 1]
l1 = [2000, 5, 1, 100, -1, -2000, 200, 4000]
count = 0
for i in range(len(l1)-1):
    for j in range(len(l1)-1-i):
        count += 1
        if l1[j] > l1[j+1]:
            l1[j], l1[j+1] = l1[j+1], l1[j]
print(l1)
print(count)
```

```python
# bubble sort for descending order
l1 = [1, 4, 5, 2, 3]

for i in range(len(l1)-1):
    for j in range(len(l1)-1-i):
        if l1[j] < l1[j+1]:
            l1[j], l1[j+1] = l1[j+1], l1[j]
l1
```

```python
# insertion sort
l1 = [23, 1, 10, 5, 2]

for i in range(1, len(l1)):
    key = l1[i]
    j = i-1
    while j>=0 and key < l1[j]:
        l1[j+1] = l1[j]
        j -= 1
    l1[j+1] = key
l1
```
