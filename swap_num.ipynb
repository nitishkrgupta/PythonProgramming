```python
# swapping two numbers
# take input of two numbers

a = int(input("a : "))
b = int(input("b : "))

print(f"Before swapping \n a : {a} \t b : {b}")
# using third variable
c = a
a = b
b = c
print(f"After swapping \n a : {a} \t b : {b}")
```

```python
# swapping of three numbers
a = int(input("a : "))
b = int(input("b : "))
c = int(input("c : "))

print(f"Before swapping \n a : {a}\t b : {b}\t c : {c}")
# using extra variable
d = a
a = c
c = b
b = d
print(f"after swapping \n a : {a}\t b : {b}\t c : {c}")
```
