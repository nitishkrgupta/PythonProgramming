```python
# printing series of even odd mix e.g. 2 1 4 3 6...
n = int(input("Enter number : "))
even = 2
odd = 1
for i in range(n):
    if i%2==0:
        print(even, end=" ")
        even += 2
    else:
        print(odd, end=" ")
        odd += 2
```

```python
# 1 5 3 7 5 9 7 11 9 13
n = 10
val = 1

for i in range(n):
    if i%2 == 0:
        print(val, end=" ")
        val += 4
    else:
        print(val, end=" ")
        val -= 2
```

```python
# fibonacci series

n = int(input("Enter number : "))
a = 0 
b = 1
print(a, end=" ")
print(b, end=" ")
for i in range(n - 2):
    c = a + b
    print(c, end=" ")
    a = b
    b = c
```
